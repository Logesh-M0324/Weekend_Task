## 8.1 Important Words

Positive reviews commonly contain sentiment-related words such as:

**Positive:** excellent, amazing, great, wonderful, enjoyable

**Negative:** terrible, boring, worst, poor, disappointing

These words provide useful sentiment signals to the models. However, sentiment cannot always be determined from individual words because context, negation, sarcasm, and the relationship between words can change the meaning of a review.

---

## 8.2 Correctly Classified Reviews

The optimized ANN model correctly classified the majority of the test reviews.

**ANN Accuracy: 90.52%**

This indicates that the model learned useful sentiment patterns from the IMDb reviews and was able to classify most previously unseen reviews correctly.

---

## 8.3 Misclassified Reviews

Some reviews were classified incorrectly even though the model achieved good overall performance.

Common reasons for misclassification include:

* **Sarcasm** – positive words may be used to express a negative opinion.
* **Mixed opinions** – a review may contain both positive and negative statements.
* **Ambiguous words** – the meaning of a word can depend on its context.
* **Negation** – words such as *not*, *never*, *neither*, and *nor* can change the sentiment.
* **Complex sentences** – long sentences may contain multiple opinions and conflicting sentiment.

For example:

> "The acting was excellent, but the story was disappointing."

The review contains both positive and negative expressions, making the overall sentiment more difficult to classify.

---

## 8.4 Positive vs Negative Language Patterns

The reviews contain different language patterns associated with positive and negative sentiment.

| Positive Words | Negative Words |
| -------------- | -------------- |
| Excellent      | Terrible       |
| Amazing        | Boring         |
| Great          | Worst          |
| Wonderful      | Poor           |
| Enjoyable      | Disappointing  |

Positive words generally express appreciation, enjoyment, or satisfaction, while negative words commonly express criticism, dissatisfaction, or disappointment.

The model learns patterns from these words along with their surrounding context to predict the sentiment of a review.

---

## 8.5 Model Limitations

Although the final model achieved good performance, some limitations remain:

* Difficulty understanding sarcasm.
* Difficulty handling mixed sentiment.
* Limited understanding of complex contextual relationships.
* Difficulty with ambiguous expressions.
* Long reviews may contain multiple or conflicting opinions.
* Model performance depends on the IMDb movie-review dataset.
* The model may not perform equally well on reviews from completely different domains.

---

## 8.6 Effect of Preprocessing and Optimization

Text preprocessing converted the raw IMDb reviews into a standardized format suitable for model training.

The preprocessing process included:

**Lowercase Conversion → HTML Removal → Character/Punctuation Handling → Stopword Handling → Tokenization → Integer Encoding → Padding**

Special attention was given to sentiment-relevant negation words such as **not, never, neither, and nor**, because removing these words can change the meaning of a review.

Model optimization further improved the performance of the ANN.

| Baseline ANN  | **87.44%** |
| Optimized ANN | **90.52%** |

The optimized ANN therefore performed better than the baseline ANN after applying the selected enhancement techniques.

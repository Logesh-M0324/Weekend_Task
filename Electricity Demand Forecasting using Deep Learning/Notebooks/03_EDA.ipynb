```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Dataset/df_for_EDA.csv")
```

```python
df
```

```python
df.index 
```

```python
df
```

```python
hourly_profile = df.groupby("Hour")["PJME_MW"].mean()

plt.figure(figsize=(12, 5))

plt.plot(
    hourly_profile.index,
    hourly_profile.values,
    marker="o"
)

plt.title("Average Hourly Electricity Demand")
plt.xlabel("Hour of Day")
plt.ylabel("Average Demand (MW)")
plt.xticks(range(24))
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
weekly_profile = df.groupby("DayOfWeek")["PJME_MW"].mean()

days = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

plt.figure(figsize=(12, 5))

plt.plot(
    days,
    weekly_profile.values,
    marker="o"
)

plt.title("Average Weekly Electricity Demand")
plt.xlabel("Day of Week")
plt.ylabel("Average Demand (MW)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
monthly_profile = df.groupby("Month")["PJME_MW"].mean()

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_profile.index,
    monthly_profile.values,
    marker="o"
)

plt.title("Average Monthly Electricity Demand")
plt.xlabel("Month")
plt.ylabel("Average Demand (MW)")
plt.xticks(range(1, 13))
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
annual_profile = df.groupby("Year")["PJME_MW"].mean()

plt.figure(figsize=(12, 5))

plt.plot(
    annual_profile.index,
    annual_profile.values,
    marker="o"
)

plt.title("Average Annual Electricity Demand")
plt.xlabel("Year")
plt.ylabel("Average Demand (MW)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
heatmap_data = df.pivot_table(
    values="PJME_MW",
    index="Hour",
    columns="DayOfWeek",
    aggfunc="mean"
)

heatmap_data.columns = days

plt.figure(figsize=(12, 8))

sns.heatmap(
    heatmap_data,
    annot=False,
    cmap="YlOrRd"
)

plt.title("Average Electricity Demand: Hour vs Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Hour")

plt.tight_layout()
plt.show()
```

```python
median_demand = df["PJME_MW"].median()

df["Demand_Period"] = np.where(
    df["PJME_MW"] >= median_demand,
    "Peak",
    "Off-Peak"
)
```

```python
df
```

```python
peak_summary = df.groupby("Demand_Period")["PJME_MW"].agg(
    ["count", "mean", "min", "max"]
)

peak_summary
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Demand_Period",
    y="PJME_MW"
)

plt.title("Peak vs Off-Peak Electricity Demand")
plt.xlabel("Demand Period")
plt.ylabel("Demand (MW)")

plt.tight_layout()
plt.show()
```

```python
sample = df.iloc[-24 * 30:]

plt.figure(figsize=(15, 6))

plt.plot(
    sample.index,
    sample["PJME_MW"],
    label="Actual Demand"
)

plt.plot(
    sample.index,
    sample["RollingMean_24"],
    label="24-Hour Rolling Mean"
)

plt.title("Electricity Demand and 24-Hour Rolling Mean")
plt.xlabel("Date")
plt.ylabel("Demand (MW)")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.plot(
    sample.index,
    sample["RollingStd_24"]
)

plt.title("24-Hour Rolling Standard Deviation")
plt.xlabel("Date")
plt.ylabel("Rolling Standard Deviation")

plt.tight_layout()
plt.show()
```

```python
lag_columns = [
    "PJME_MW",
    "Lag_1",
    "Lag_24",
    "Lag_168"
]

lag_correlation = df[lag_columns].corr()

lag_correlation
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    lag_correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Between Demand and Lag Features")

plt.tight_layout()
plt.show()
```

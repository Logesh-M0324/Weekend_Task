```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

import warnings
warnings.filterwarnings("ignore")
```

```python
PJME_data["Datetime"] = pd.to_datetime(PJM_data["Datetime"])
```

```python
PJME_data.info()
```

```python
PJME_data.sort_values("Datetime")
```

```python
PJME_data["Datetime"]
```

```python
PJME_data["Datetime"].diff()
```

```python
PJME_data["Datetime"].diff().value_counts()
```

```python
full_range = pd.date_range( start=PJME_data["Datetime"].min(), end=PJME_data["Datetime"].max(), freq="h" )
```

```python
full_range
```

```python
missing_timestamps = full_range.difference(PJME_data["Datetime"])

len(missing_timestamps)
```

```python
missing_timestamps
```

```python
plt.figure(figsize=(15,5))

plt.plot(
    PJME_data["Datetime"],
    PJME_data["PJME_MW"]
)

plt.xlabel("Date")
plt.ylabel("Electricity Demand (MW)")
plt.title("Hourly Electricity Demand")

plt.show()
```

```python
PJME_data["Hour"] = PJME_data["Datetime"].dt.hour

hourly_profile = PJME_data.groupby("Hour")["PJME_MW"].mean()
```

```python
hourly_profile
```

```python
plt.figure(figsize=(10,5))

hourly_profile.plot()

plt.title("Average Hourly Electricity Demand")
plt.xlabel("Hour")
plt.ylabel("Average Demand (MW)")

plt.show()
```

# Weekly Pattern

```python
PJME_data["DayOfWeek"] = PJME_data["Datetime"].dt.dayofweek

weekly_profile = PJME_data.groupby("DayOfWeek")["PJME_MW"].mean()
```

```python
weekly_profile
```

```python
plt.figure(figsize=(10,5))

weekly_profile.plot(kind="bar")

plt.title("Average Electricity Demand by Day of Week")
plt.xlabel("Day")
plt.ylabel("Average Demand (MW)")

plt.show()
```

# Monthly Demand

```python
PJME_data["Month"] = PJME_data["Datetime"].dt.month

monthly_profile = PJME_data.groupby("Month")["PJME_MW"].mean()
```

```python
plt.figure(figsize=(10,5))

monthly_profile.plot(marker="o")

plt.title("Monthly Electricity Demand")
plt.xlabel("Month")
plt.ylabel("Average Demand (MW)")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    PJME_data["PJME_MW"],
    kde=True
)

plt.title("Electricity Demand Distribution")

plt.show()
```
